````python

import os
import json
import re
import torch
from huggingface_hub import hf_hub_download

DATASET_ID = "rakesh-bishnoi29/structuraai-dataset"
MODEL_ID = "CohereLabs/aya-expanse-8b"

def load_jsonl_from_hf(filename):
    path = hf_hub_download(
        repo_id=DATASET_ID,
        filename=filename,
        repo_type="dataset"
    )

    with open(path, "r", encoding="utf-8") as f:
        return [json.loads(line) for line in f if line.strip()]

train_data = load_jsonl_from_hf("train.jsonl")
validation_data = load_jsonl_from_hf("validation.jsonl")
test_data = load_jsonl_from_hf("test.jsonl")

def extract_ocr_text(record):
    marker = "OCR TEXT:"
    original = record["input"]

    if marker not in original:
        raise ValueError(
            f"OCR TEXT marker missing: {record['document_id']}"
        )

    return original.split(marker, 1)[1].strip()

baseline_inputs = [
    {
        "document_id": record["document_id"],
        "ocr_text": extract_ocr_text(record)
    }
    for record in test_data
]

EXTRACTION_INSTRUCTION = """
You are an intelligent invoice information extraction assistant.

Carefully analyze the invoice OCR text.

Identify all relevant information contained in the invoice.
Understand the relationships between the supplier, customer,
invoice details, dates, line items, taxes, totals, and payment
information.

Independently organize the extracted information into a
logical, well-structured JSON object.

You are not required to follow any predefined JSON schema.
Choose clear and meaningful field names and appropriate
nested objects and arrays.

Rules:
1. Extract only information supported by the OCR text.
2. Do not invent or assume missing values.
3. Use null for unavailable or unclear values.
4. Preserve numerical values accurately.
5. Represent multiple invoice items as an array.
6. Return one valid JSON object.
7. Do not include explanations or Markdown code fences.
8. Return JSON only.
"""

def build_prompt(ocr_text):
    return (
        EXTRACTION_INSTRUCTION
        + "\n\nINVOICE OCR TEXT:\n\n"
        + ocr_text
        + "\n\nJSON OUTPUT:"
    )

assert len(train_data) == 196
assert len(validation_data) == 42
assert len(test_data) == 42

print("Training:", len(train_data))
print("Validation:", len(validation_data))
print("Testing:", len(test_data))
print("Prepared OCR inputs:", len(baseline_inputs))
````

````python
import torch
import json
import re
import os
from datetime import datetime, timezone

# Select first test invoice
sample = baseline_inputs[0]

# Build the same schema-free prompt
prompt = build_prompt(sample["ocr_text"])

messages = [
    {"role": "user", "content": prompt}
]

inputs = tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    return_dict=True,
    return_tensors="pt"
)

inputs = inputs.to(model.device)

input_tokens = inputs["input_ids"].shape[1]
max_new_tokens = 1800

print("Document:", sample["document_id"])
print("Input tokens:", input_tokens)
print("Context limit:", model.config.max_position_embeddings)

assert (
    input_tokens + max_new_tokens
    <= model.config.max_position_embeddings
), "Input plus generation tokens exceed context limit."

# Generate prediction
start_time = datetime.now(timezone.utc)

with torch.inference_mode():
    outputs = model.generate(
        **inputs,
        max_new_tokens=max_new_tokens,
        do_sample=False,
        num_beams=1,
        pad_token_id=(
            tokenizer.pad_token_id
            if tokenizer.pad_token_id is not None
            else tokenizer.eos_token_id
        ),
        eos_token_id=tokenizer.eos_token_id
    )

torch.cuda.synchronize()

elapsed = (
    datetime.now(timezone.utc) - start_time
).total_seconds()

# Decode generated response
generated_tokens = outputs[0][input_tokens:]

raw_prediction = tokenizer.decode(
    generated_tokens,
    skip_special_tokens=True
).strip()

print("\n========== AYA RESPONSE ==========\n")
print(raw_prediction)

print("\nGeneration time:", round(elapsed, 2), "seconds")
print("Generated tokens:", len(generated_tokens))
````

````python

import os
import json
import re
import torch
from datetime import datetime, timezone

# Output files
PREDICTIONS_PATH = os.path.join(RUN_DIR, "baseline_predictions.jsonl")
LOG_PATH = os.path.join(RUN_DIR, "baseline_run_log.json")

MAX_NEW_TOKENS = 1800

# Resume support
completed_ids = set()

if os.path.exists(PREDICTIONS_PATH):
    with open(PREDICTIONS_PATH, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                record = json.loads(line)
                completed_ids.add(record["document_id"])

print("Total test invoices:", len(baseline_inputs))
print("Already completed:", len(completed_ids))
print("Remaining:", len(baseline_inputs) - len(completed_ids))

run_start = datetime.now(timezone.utc)
success_count = 0
failure_count = 0

for index, sample in enumerate(baseline_inputs, start=1):

    doc_id = sample["document_id"]

    if doc_id in completed_ids:
        continue

    print(f"\n[{index}/{len(baseline_inputs)}] Processing {doc_id}")

    try:
        prompt = build_prompt(sample["ocr_text"])

        messages = [{"role": "user", "content": prompt}]

        inputs = tokenizer.apply_chat_template(
            messages,
            tokenize=True,
            add_generation_prompt=True,
            return_dict=True,
            return_tensors="pt"
        ).to(model.device)

        input_tokens = inputs["input_ids"].shape[1]

        # Ensure input + generation fit model context
        available_tokens = (
            model.config.max_position_embeddings - input_tokens
        )

        if available_tokens <= 0:
            raise ValueError("Input exceeds model context limit.")

        generation_limit = min(
            MAX_NEW_TOKENS,
            available_tokens
        )

        start_time = datetime.now(timezone.utc)

        with torch.inference_mode():
            outputs = model.generate(
                **inputs,
                max_new_tokens=generation_limit,
                do_sample=False,
                num_beams=1,
                pad_token_id=(
                    tokenizer.pad_token_id
                    if tokenizer.pad_token_id is not None
                    else tokenizer.eos_token_id
                ),
                eos_token_id=tokenizer.eos_token_id
            )

        torch.cuda.synchronize()

        elapsed = (
            datetime.now(timezone.utc) - start_time
        ).total_seconds()

        generated_tokens = outputs[0][input_tokens:]

        raw_output = tokenizer.decode(
            generated_tokens,
            skip_special_tokens=True
        ).strip()

        cleaned = re.sub(
            r"^```(?:json)?\s*|\s*```$",
            "",
            raw_output
        )

        try:
            parsed_output = json.loads(cleaned)
            json_valid = True
            parse_error = None
        except json.JSONDecodeError as error:
            parsed_output = None
            json_valid = False
            parse_error = str(error)

        record = {
            "document_id": doc_id,
            "raw_model_output": raw_output,
            "parsed_prediction": parsed_output,
            "json_parseable": json_valid,
            "parse_error": parse_error,
            "input_tokens": input_tokens,
            "generated_tokens": len(generated_tokens),
            "generation_limit": generation_limit,
            "hit_token_limit": len(generated_tokens) >= generation_limit,
            "inference_seconds": round(elapsed, 3),
            "model": MODEL_ID,
            "experiment": "Experiment A - Schema-free"
        }

        with open(PREDICTIONS_PATH, "a", encoding="utf-8") as f:
            f.write(json.dumps(record, ensure_ascii=False) + "\n")

        success_count += 1

        print(
            f"JSON valid: {json_valid} | "
            f"Tokens: {len(generated_tokens)} | "
            f"Time: {elapsed:.2f}s"
        )

    except Exception as error:
        failure_count += 1
        print("ERROR:", repr(error))

        # Save error separately
        error_path = os.path.join(RUN_DIR, "baseline_errors.jsonl")

        with open(error_path, "a", encoding="utf-8") as f:
            f.write(json.dumps({
                "document_id": doc_id,
                "error": repr(error)
            }) + "\n")

        if torch.cuda.is_available():
            torch.cuda.empty_cache()

run_end = datetime.now(timezone.utc)

# Run summary
summary = {
    "run_id": RUN_ID,
    "model": MODEL_ID,
    "experiment": "Experiment A - Schema-free",
    "total_test_documents": len(baseline_inputs),
    "records_already_completed": len(completed_ids),
    "new_successes": success_count,
    "new_failures": failure_count,
    "run_started_utc": run_start.isoformat(),
    "run_finished_utc": run_end.isoformat(),
    "predictions_file": PREDICTIONS_PATH
}

with open(LOG_PATH, "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2)

print("\n========== BASELINE RUN SUMMARY ==========")
print(json.dumps(summary, indent=2))
````

````python

import os
import json
import statistics

PREDICTIONS_PATH = os.path.join(
    RUN_DIR,
    "baseline_predictions.jsonl"
)

records = []

with open(PREDICTIONS_PATH, "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():
            records.append(json.loads(line))

total = len(records)
unique_ids = len(set(r["document_id"] for r in records))

valid_json = sum(
    r["json_parseable"] for r in records
)

invalid_json = total - valid_json

token_limited = sum(
    r["hit_token_limit"] for r in records
)

times = [
    r["inference_seconds"]
    for r in records
    if r.get("inference_seconds") is not None
]

generated_tokens = [
    r["generated_tokens"]
    for r in records
    if r.get("generated_tokens") is not None
]

print("========== BASELINE PREDICTION AUDIT ==========")
print("Total records:", total)
print("Unique document IDs:", unique_ids)
print("Valid JSON:", valid_json)
print("Invalid JSON:", invalid_json)
print("JSON parse rate:", round(valid_json / total * 100, 2), "%")
print("Generation limit reached:", token_limited)

if times:
    print("Average inference time:", round(statistics.mean(times), 2), "seconds")
    print("Total inference time:", round(sum(times), 2), "seconds")

if generated_tokens:
    print("Average generated tokens:", round(statistics.mean(generated_tokens), 2))

# Verify against expected test-set size
assert total == 42, f"Expected 42 records, found {total}"
assert unique_ids == 42, "Duplicate or missing document IDs detected"

print("\nRecord count and uniqueness checks passed.")
````
